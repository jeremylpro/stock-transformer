# 03 - Feature diagnostics

Check feature coverage, univariate relationships with the next-day label, and the saved combined-model importances.

```python
import json
import pandas as pd

dataset = pd.read_parquet('../data/processed/dataset.parquet')
feature_cols = [c for c in dataset.columns if c.startswith(('ret_', 'vol_', 'ta_', 'sent_'))]
feature_cols
```

```python
dataset[feature_cols].isna().mean().sort_values(ascending=False)
```

```python
# Point-biserial correlation with the next-day direction (diagnostic only).
corr = dataset[feature_cols + ['y_next_up']].corr()['y_next_up'].drop('y_next_up')
corr.sort_values()
```

```python
metrics = json.load(open('../reports/metrics.json'))
pd.DataFrame(metrics['models']['combined_xgb']['feature_importance']).head(15)
```

```python
pd.DataFrame(
    {k: v['test'] for k, v in metrics['models'].items()}
).T[['auc', 'accuracy', 'brier', 'log_loss']]
```
